```python
input_string= "Python"
reverse_string= ""
for char in input_string:
    reverse_string = char + reverse_string
print(reverse_string)
   
```

```python
def reverse_string(input_string: str)->str:
    """
    Function is used to reverse a given string using for loop.
    Args:
        input_string(str)--> The string to reverse which a user enters.
    Return:
        str--> Return the entered string in reverse order.
    Raises:
        TypeError--> If input is not a string    
    """
    # validate the input type
    if not isinstance(input_string, str):
        raise TypeError("Input must be a string")
        
    reversed_text= ""
    for char in input_string:
        reversed_text = char + reversed_text
    return reversed_text
# main execution
if __name__ == "__main__":
    try:
        # take input from the user
        user_input= input("Enter the string to reverse: ")
        # call function
        result= reverse_string(user_input)
        # display reversed string
        print(f"Reversed string: {result}")
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)    
```
