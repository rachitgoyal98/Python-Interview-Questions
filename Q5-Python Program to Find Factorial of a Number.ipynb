```python
def factorial_finder(num:int)-> int:
    """
    Function to find factorial of a given positive integer.
    Args:
        num(int): number to find factorial.
    Returns:
        int: factorial of the provided number.
    Raises:
        ValueError: if number is negative
        TypeError: if number is not an integer
    """

    # validate the number is not a boolean
    if isinstance(num, bool):
        raise TypeError("Boolean values are not allowed")
    # validate number type
    if not isinstance(num, int):
        raise TypeError("Given number must be an integer")
    # validate number is positive
    if num<0:
        raise ValueError("Given number must be positive")

    factorial = 1
    for i in range(1,num+1):
        factorial= factorial*i
    return factorial

# main execution
if __name__ == "__main__":
    try:
        num = input("Enter your number to find factorial: ")
        num = int(num)
        output = factorial_finder(num)
        print(f"Factorial of {num} is: ", output)
    except ValueError as e:
        print("Input must be a positive integer number")
    except TypeError as e:
        print(e)
    
```
