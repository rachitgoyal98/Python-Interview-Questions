```python
from typing import List
def second_largest(nums: List[int]) -> int:
    """
    Function returns second largest number in a list.
    Args:
        nums(List[int]): A list of integers.
    Returns:
        int: Second largest value.
    Raises:
        TypeError: If input is not a list or values are not integer.
        ValueError: If the list has fewer than two unique elements.
    """

    # validate the input must be a list
    if not isinstance(nums, List):
        raise TypeError("Input must be a list of integer")
    # validate all elements in a list are intigers
    if any(not isinstance(current_number,int) for current_number in nums):
        raise TypeError("All elements in a list should be integer")
    # check list has atleast two elements
    if len(nums) < 2:
        raise ValueError("List should have atlease two elements")

    if nums[0] > nums[1]:
        largest = nums[0]
        second = nums[1]
    else:
        largest = nums[1]
        second = nums[0]

    for current_number in nums[2:]:
        if current_number > largest:
            second = largest
            largest = current_number
    
        elif current_number != largest and current_number > second:
            second = current_number

    if second == largest:
        raise ValueError("List must contain atleast two unique elements")
    
    return second

# ----main execution----
if __name__ == "__main__":
    numbers_list = [1,2,34,54,22,3,44,5,66,7,455,4,3,4,56,65]
    try:
        result = second_largest(numbers_list)
    except (TypeError, ValueError) as error:
        print("Error: ", error)
    except Exception as e:
        print("Error: ", e)
    else:
        print("second_largest: ", result)

    
```
