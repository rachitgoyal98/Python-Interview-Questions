```python
def is_prime_number(number: int)->bool:
    """
    Function to check whether a number is a prime or not.
    Args:
        number(int): Number to check.
    Returns: 
        bool: True- if the number is prime, False- if the number is not a prime
    Raises:
        TypeError: If input is not an integer.
        ValueError: If number is less than 2, since prime numbers are greater than or equals 2.
    """

    # validate the input is an integer or not
    if not isinstance(number, int):
        raise TypeError("The number must be an integer")
    # validate that the number should greater than or equals to 2
    if not number> 2 or number==2:
        raise ValueError("The number must be greater than or equals to 2")

    for i in range(2,int(number**.5+1)):
        if number % i == 0:
            return False
            break
    else:
        return True

# main execution
if __name__ == "__main__":
    try:
        # take input from user
        user_input = input("Enter a number to check if it is a prime or not: ")
        # validate and convert input to int
        if not user_input.strip().isdigit():
            raise TypeError("Input must be a digit (no letters, no string and no float)")
        number = int(user_input)
        # calling the function
        if is_prime_number(number):
            print(f"{number} is a prime number.")
        else:
            print(f"{number} is not a prime number.")
    except TypeError as e:
        print(e)
    except ValueError as e:
        print(e)
    except exception as e:
        print(e)
```
