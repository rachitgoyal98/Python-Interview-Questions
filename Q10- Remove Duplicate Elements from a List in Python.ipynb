```python

from typing import List, Any

def remove_duplicate(input_list: List[Any]) -> List[Any]:
    """
    Function is used to remove duplicate values from a list.

    Args:
        input_list: A list containing any type of element.

    Returns:
        A list containing unique values.

    Raises:
        TypeError: If input is not a list.
        ValueError: If input is empty.
    """

    # Validate input is a list
    if not isinstance(input_list, list):
        raise TypeError("Input must be a list")

    # Validate input is not empty
    if len(input_list) == 0:
        raise ValueError("Input list must not be empty")

    # Check if list has only one element
    if len(input_list) == 1:
        print("List has only one element. No duplicates to remove")
        return input_list

    result = []

    for i in input_list:
        if i not in result:
            result.append(i)

    return result


# Main execution
if __name__ == "__main__":
    try:
        number_list = [1, 2, 3, 4, 5, 3, 4, 5, 2, 2, 3, 4, 5, 6, 7, 6, 7, 7, 8]

        result = remove_duplicate(number_list)

    except (TypeError, ValueError) as error:
        print("Error:", error)

    except Exception as error:
        print("Error:", error)

    else:
        print("After removing duplicates:", result)
```
