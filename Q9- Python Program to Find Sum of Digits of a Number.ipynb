```python
def sum_of_digits(n: int)-> int:
    """
    Function calculates the sum of digits of a non-negative integer.
    Args:
        n(int): number of which sum needs to be find.
    Returns:
        int: the sum of digits.
    Raises:
        TypeError: if input is not an integer.
        ValueError: if input is negative
    """

    if not isinstance(n, int) or isinstance(n, bool):
        raise TypeError("Input must be an integer")
    if n < 0:
        raise ValueError("Input must be a non-negative integer")

    total = 0
    while n > 0:
        total = total + (n%10)
        n = n // 10
    return total
#----main execution----

if __name__ == "__main__":
    num = input("Enter your number: ")

    try:
        num = int(num)
    except ValueError:
        print("Error: Please enter valid non-negative integer only")
    
    else:
        try:
            result = sum_of_digits(num)
            print(f"Sum of digits is: {result}")
        except (TypeError, ValueError) as error:
            print("Error:", error)
    
```
