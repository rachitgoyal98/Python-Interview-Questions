```python
def find_largest_number(number_list: list[float])-> float:
    """
    Function to find largest number in the list
    Args:
        number_list(number_list: list[float])-> A list containing numeric values (int of float).
    Returns:
        Largest value in the list
    Raises:
        TypeError: If input is not a list or non-numeric value.
        ValueError: If the list is empty.   
    """

    # validate that the input is a list
    if not isinstance(number_list,list):
        raise TypeError("Input must be a list of numbers (int or float")
    # validate that input is not empty
    if not number_list:
        raise ValueError("List cannot be empty")
    # validte that list elements should be numeric
    for num in number_list:
        if not isinstance(num,(int,float)):
            raise TypeError("All elements must be numeric")

    largest = number_list[0]
    for num in number_list:
        if num>largest:
            largest = num
    return largest
    
# main execution
if __name__ == "__main__":
    try:
        # take input from the user
        user_input= input("Enter numbers in the list separated by space int or float only: ")
        # converting number_list into list
        number_list= [float(item) for item in user_input.split()] 
        # call the function
        result = find_largest_number(number_list)
        #display the result
        print(f"The largest number in the list is: {result}")
    except (TypeError, ValueError) as error: 
        print(error)
    except Exception as e:
        print(e)
        
```
